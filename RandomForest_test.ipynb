```python
# ============================================================
# LOAD PRETRAINED MODEL
# ============================================================

MODEL_PATH = "wellness_randomforest_pipeline.pkl"

model_package = joblib.load(MODEL_PATH)

model = model_package["model"]
FEATURE_COLUMNS = model_package["features"]
RISK_MAPPING = model_package["risk_mapping"]

print("Pretrained Random Forest loaded successfully!")

print("\nNumber of features:", len(FEATURE_COLUMNS))

print("\nFeatures used by the model:")
for i, feature in enumerate(FEATURE_COLUMNS, 1):
    print(f"{i}. {feature}")
```

```python
def predict_student_risk(student_row):
    """
    Predict risk level for a single student using the pretrained Random Forest model.
    """

    # Convert student data into DataFrame
    input_df = pd.DataFrame([student_row])

    # Select only the features used during training
    input_df = input_df[FEATURE_COLUMNS]

    # Prediction
    predicted_class = model_package["model"].predict(input_df)[0]

    # Probability
    probabilities = model_package["model"].predict_proba(input_df)[0]

    # Convert class number to risk name
    risk_mapping = model_package["risk_mapping"]
    predicted_risk = risk_mapping[predicted_class]

    result = {
        "predicted_risk": predicted_risk,
        "low_probability": probabilities[0],
        "medium_probability": probabilities[1],
        "high_probability": probabilities[2]
    }

    return result
```

```python
def predict_student_risk(student_row):

    input_df = pd.DataFrame([student_row])

    # Select training features
    input_df = input_df[FEATURE_COLUMNS]

    # Get model
    model = model_package["model"]

    # Prediction
    predicted_class = model.predict(input_df)[0]

    # Probabilities
    probabilities = model.predict_proba(input_df)[0]

    # Risk mapping
    risk_mapping = model_package["risk_mapping"]

    predicted_risk = risk_mapping[predicted_class]

    return {
        "Predicted Risk": predicted_risk,
        "Low Probability (%)": round(probabilities[0] * 100, 2),
        "Medium Probability (%)": round(probabilities[1] * 100, 2),
        "High Probability (%)": round(probabilities[2] * 100, 2)
    }
```

```python
def predict_student_risk(student_row):
    input_df = pd.DataFrame([student_row])
    input_df = input_df[FEATURE_COLUMNS]

    model = model_package["model"]

    predicted_class = model.predict(input_df)[0]
    probabilities = model.predict_proba(input_df)[0]

    risk_mapping = model_package["risk_mapping"]
    predicted_risk = risk_mapping[predicted_class]

    return (
        predicted_risk,
        probabilities[0],
        probabilities[1],
        probabilities[2]
    )
```
