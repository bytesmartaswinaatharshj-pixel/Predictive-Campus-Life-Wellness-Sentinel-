# Predictive Campus Life Wellness Sentinel
## Random Forest - Dataset Generator

### Objective
This notebook generates a synthetic student behavioral dataset for the
Predictive Campus Life Wellness Sentinel project.

The dataset represents weekly campus engagement and behavioral indicators
for students and is designed for developing an early-warning risk
prediction prototype.

### Dataset Details
- Number of students: 1,000
- Number of weeks: 12
- Data type: Synthetic
- Target: Low / Medium / High behavioral risk

### Main Activities
1. Generate synthetic student behavioral data
2. Perform feature engineering
3. Generate future behavioral indicators
4. Create the risk labels
5. Remove future information from ML inputs
6. Save the final dataset as a CSV file


```python
# ============================================================
# NOTEBOOK 1
# RandomForest_dataset_generator
# Predictive Campus Life Wellness Sentinel
# ============================================================

# Import essential libraries for data handling and system operations
import os
import numpy as np
import pandas as pd

# Set a random seed to ensure reproducible results across runs
np.random.seed(42)

# ------------------------------------------------------------
# CONFIGURATION
# ------------------------------------------------------------

# Define global constants for dataset dimensions
NUM_STUDENTS = 1000  # Total number of unique students
NUM_WEEKS = 12  # Total weeks of tracking per student

# Output file path where the final CSV will be stored
OUTPUT_PATH = "/content/student_wellness_prediction_dataset.csv"


# ============================================================
# DATA GENERATION FUNCTION
# ============================================================


def generate_wellness_dataset(
    num_students=NUM_STUDENTS, num_weeks=NUM_WEEKS
):
  # Initialize an empty list to accumulate individual weekly records
  records = []

  # Iterate through each student (assigning IDs from 10001 to 11000)
  for student_id in range(10001, 10001 + num_students):

    # Establish a baseline engagement level for the student (random float between 45 and 85)
    base_engagement = np.random.uniform(45, 85)

    # Generate sequential data for weeks 1 through 12
    for week in range(1, num_weeks + 1):

      # Introduce random normal weekly variance around baseline (mean=0, std=5)
      weekly_change = np.random.normal(0, 5)

      # Generate synthetic metric: Facility usage (bounded between 0 and 100)
      facility_usage = np.clip(
          base_engagement * 0.8 + weekly_change + np.random.normal(0, 8),
          0,
          100,
      )

      # Generate synthetic metric: Dining activity (bounded between 0 and 100)
      dining_activity = np.clip(
          base_engagement * 0.7 + np.random.normal(0, 10), 0, 100
      )

      # Generate Poisson-distributed count for campus event participation (max 10)
      event_participation = np.clip(
          np.random.poisson(max(base_engagement / 30, 0.5)), 0, 10
      )

      # Generate Poisson-distributed count for club participation (max 10)
      club_participation = np.clip(
          np.random.poisson(max(base_engagement / 35, 0.5)), 0, 10
      )

      # Generate synthetic metric: Residence engagement (bounded between 0 and 100)
      residence_engagement = np.clip(
          base_engagement * 0.6 + np.random.normal(0, 12), 0, 100
      )

      # Generate synthetic metric: Recreation activity (bounded between 0 and 100)
      recreation_activity = np.clip(
          base_engagement * 0.65 + np.random.normal(0, 12), 0, 100
      )

      # Generate synthetic metric: Social interactions count (bounded between 0 and 30)
      social_interactions = np.clip(
          base_engagement * 0.25 + np.random.normal(0, 5), 0, 30
      )

      # Generate synthetic metric: Communication activity (bounded between 0 and 40)
      communication_activity = np.clip(
          base_engagement * 0.3 + np.random.normal(0, 6), 0, 40
      )

      # Generate synthetic metric: Sleep quality score (bounded between 0 and 100)
      sleep_quality = np.clip(
          base_engagement * 0.7 + np.random.normal(0, 10), 0, 100
      )

      # Generate synthetic metric: Academic engagement score (bounded between 0 and 100)
      academic_engagement = np.clip(
          base_engagement * 0.75 + np.random.normal(0, 10), 0, 100
      )

      # ------------------------------------------------
      # CAMPUS ENGAGEMENT SCORE
      # ------------------------------------------------

      # Compute composite campus engagement score as the mean of weighted indicators
      campus_engagement_score = np.mean([
          facility_usage,
          dining_activity,
          residence_engagement,
          recreation_activity,
          academic_engagement,
          social_interactions * 3,  # Scale weight
          communication_activity * 2,  # Scale weight
      ])

      # Ensure overall engagement score stays within strict range [0, 100]
      campus_engagement_score = np.clip(campus_engagement_score, 0, 100)

      # ------------------------------------------------
      # SOCIAL ISOLATION SCORE
      # ------------------------------------------------

      # Calculate isolation score inversely from social and activity participation
      social_isolation_score = np.clip(
          100
          - np.mean([
              social_interactions * 3,
              communication_activity * 2,
              event_participation * 8,
              club_participation * 8,
              residence_engagement,
          ]),
          0,
          100,
      )

      # Append all parameters for current student-week record to array
      records.append({
          "student_id": student_id,
          "week": week,
          "facility_usage": facility_usage,
          "dining_activity": dining_activity,
          "event_participation": event_participation,
          "club_participation": club_participation,
          "residence_engagement": residence_engagement,
          "recreation_activity": recreation_activity,
          "social_interactions": social_interactions,
          "communication_activity": communication_activity,
          "sleep_quality": sleep_quality,
          "academic_engagement": academic_engagement,
          "campus_engagement_score": campus_engagement_score,
          "social_isolation_score": social_isolation_score,
      })

  # Return complete records array formatted as a Pandas DataFrame
  return pd.DataFrame(records)


# ============================================================
# GENERATE DATA
# ============================================================

# Execute data generator function
df = generate_wellness_dataset()

# Print confirmation messages and dataframe dimensions
print("Dataset generated successfully.")
print("Shape:", df.shape)

# Display the first 5 rows of the generated dataset in output view
display(df.head())
```

```python
# ============================================================
# TEMPORAL FEATURE ENGINEERING
# ============================================================

# Sort dataframe chronologically by student and week, then reindex
df = df.sort_values(["student_id", "week"]).reset_index(drop=True)

# Shift engagement score down by 1 week per student to capture prior week performance
df["previous_engagement_score"] = df.groupby("student_id")[
    "campus_engagement_score"
].shift(1)

# Calculate percentage change between current week engagement and prior week engagement
df["engagement_change_pct"] = (
    (df["campus_engagement_score"] - df["previous_engagement_score"])
    / df["previous_engagement_score"]
) * 100

# Compute a 3-week rolling average of engagement score per student
df["rolling_3_week_engagement"] = df.groupby("student_id")[
    "campus_engagement_score"
].transform(lambda x: x.rolling(window=3, min_periods=1).mean())

# Shift engagement score up by 1 week (future target preview for risk detection)
df["next_week_engagement"] = df.groupby("student_id")[
    "campus_engagement_score"
].shift(-1)

# Shift isolation score up by 1 week (future target preview for risk detection)
df["next_week_isolation"] = df.groupby("student_id")[
    "social_isolation_score"
].shift(-1)

# Calculate projected percentage change into the next upcoming week
df["future_engagement_change_pct"] = (
    (df["next_week_engagement"] - df["campus_engagement_score"])
    / df["campus_engagement_score"]
) * 100
```
